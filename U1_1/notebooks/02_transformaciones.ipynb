```python
import pandas as pd
import requests

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head(10)
```

```python
df.rename(columns={"entity":"country"}, inplace=True)
```

```python
df.head(10)
```

```python
df.rename(columns={'country': 'país', 'year': 'año', 'code': 'código', 'life_expectancy_0': 'esperanza_de_vida'}, inplace=True)
```

```python
df.head(10)
```

```python
df.set_index("país", inplace=True)
```

```python
df.loc[df["país"] == "Mexico"]
```

```python
df.dropna(subset=["código"], inplace=True)
df.isnull().sum()
```

```python
resultado = (
    df[df["año"] == 2020].groupby("país") ["esperanza_de_vida"].mean().sort_values(ascending=True)
)
resultado
```
